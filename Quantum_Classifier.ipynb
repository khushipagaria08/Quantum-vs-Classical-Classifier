```python
#classical ML
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

#Quantum ML
from qiskit.circuit.library import z_feature_map
from qiskit.circuit.library import RealAmplitudes

from qiskit_machine_learning.algorithms import VQC
from qiskit.primitives import StatevectorSampler
from qiskit_algorithms.optimizers import COBYLA
```

```python
iris= load_iris()

X= iris.data
y= iris.target

print(X.shape)
print(y.shape)
```

```python
X_train, X_test, y_train, y_test= train_test_split(

    X,
    y,

    test_size= 0.3,
    random_state=57
)

print(X_train.shape)
print(X_test.shape)
```

```python
scaler= MinMaxScaler()                  #quantum rotation angles work better when values are bw [0,1]
X_train= scaler.fit_transform(X_train)
X_test= scaler.fit_transform(X_test)
```

```python
#Classical Model
model= LogisticRegression(max_iter=300)
model.fit(X_train, y_train)
predictions = model.predict(X_test)
accuracy= accuracy_score(y_test, predictions)

print("Clasical Accuracy: ")
print(accuracy)
```

```python
feature_map = z_feature_map(feature_dimension=4,reps=2)

feature_map.draw("text")
```

```python
ansatz= RealAmplitudes(num_qubits= 4, reps=2)
ansatz.draw("text")
```

```python
optimizer= COBYLA(maxiter= 200)
sampler= StatevectorSampler()

vqc= VQC(
    sampler= sampler,
    feature_map= feature_map,
    ansatz=ansatz,
    optimizer=optimizer
)
```

```python
vqc.fit(X_train, y_train)
```

```python
quantum_predictions = vqc.predict(X_test)

quantum_accuracy = accuracy_score(y_test,quantum_predictions)

print("Quantum Accuracy:")
print(quantum_accuracy)
```

```python
print("\nFINAL RESULTS\n")

print(
    f"Classical Accuracy: {accuracy:.3f}"
)

print(
    f"Quantum Accuracy: {quantum_accuracy:.3f}"
)
```
